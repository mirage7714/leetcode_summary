# 本題要找出出現三次以上且間距都相同的數字共有幾個 (\#Array )
原題目連結: https://leetcode.com/problems/count-values-with-equally-spaced-occurrences-ii/description/  

You are given an integer array nums.  
An integer x is called special if:  
- x appears at least three times in nums.
- All occurrences of x are equally spaced in nums. In other words, if all occurrences of x are at indices i1 < i2 < ... < im, then i2 - i1 = i3 - i2 = ... = im - im-1.

Return the number of distinct special integers in nums.  

範例:  
Ex1:  
Input: nums = [1,8,1,5,1,5,8,5]  
Output: 2  
Explanation:  
- 1 is special because it occurs at equally spaced indices 0, 2, and 4.
- 5 is special because it occurs at equally spaced indices 3, 5, and 7.
- 8 is not special because it occurs only twice.

Therefore, the answer is 2.  

Ex2:  
Input: nums = [8,8,8,8]  
Output: 1  
Explanation:  
8 is special because it occurs at equally spaced indices 0, 1, 2, and 3. Therefore, the answer is 1.  

Ex3:  
Input: nums = [8,6,6,8,8]  
Output: 0  
Explanation:  
8 occurs at indices 0, 3, and 4, which are not equally spaced. 6 occurs only twice. Therefore, no integer is special.

* 解題想法:  
這題與No4048解法類似，首先建立以下的資料結構:  
 - count: 用來存放每個數字出現的index
 - d: 用來放每個數字的間距
 - unmatch:用來存放不符合條件的數字有哪些

   接著用迴圈開始檢查該數字是否有出現在unmatch中，如果沒有出現過則接下來檢查是否有出現在count中，如果沒出現過則在count中新增一個空的list，接著檢查在d中是否有之前的間距資料，如果沒有則看墓是否有前一個數字可計算間距並更新間距，如果有的話則檢查是否符合間距，如果間距不符合則將該數放入unmatch中，最後將該數放入count中，迴圈結束後檢查count中的key是否有出現在unmatch中且出現次數為三次以上，統計這些數字就是答案

```python
class Solution:
    def countSpecialIntegers(self, nums: list[int]) -> int:
        res = 0
        count = {}
        unmatch = set()
        d = {}
        for i in range(len(nums)):
            n = nums[i]
            if n not in unmatch:
                if n not in count.keys():
                    count[n] = []
                if count[n]:
                    if n not in d.keys():
                        d[n] = i - count[n][-1]
                    elif d[n] != i - count[n][-1]:
                        unmatch.add(n)
                count[n].append(i)
        for key, val in count.items():
            if key not in unmatch and len(val) >= 3:
                res += 1
        return res
```
