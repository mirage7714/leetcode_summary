# 本題要找出最大的整數為多少 (\#Math )
原題目連結: https://leetcode.com/problems/largest-integer-with-given-digit-sum/description/  

You are given two non-negative integers n and s.  
Return the largest integer that has at most n digits and whose sum of digits is s. If no such integer exists, return -1.  

範例:  
Ex1:  
Input: n = 2, s = 9  
Output: 90  
Explanation:  
The largest integer with at most 2 digits that has a sum of digits of 9 is 90.  

Ex2:  
Input: n = 2, s = 19  
Output: -1  
Explanation:  
There is no integer with at most 2 digits that has a sum of digits of 19, so the answer is -1.  

Ex3:  
Input: n = 5, s = 0  
Output: 0  
Explanation:  
The only non-negative integer whose digits sum to 0 is 0.

* 解題想法:  
用一個範圍為n的迴圈，如果目前的s大於或等於9的話則減去9，否則減去目前的s，等迴圈結束後檢查s是否為0，如果不為0則回傳-1，如果為0則將結果轉換為整束後回傳  

```python
class Solution:
    def largestInteger(self, n: int, s: int) -> int:
        res = ''
        for i in range(n):
            if s >= 9:
                res += '9'
                s -= 9
            else:
                res += str(s)
                s = 0
        if s > 0:
            return -1
        else:
            return int(res)
```
