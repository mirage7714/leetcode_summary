# 本題要找出出現三次且出現的間距相同的數字有多少 (\#Array )
原題目連結: https://leetcode.com/problems/count-values-with-equally-spaced-occurrences-i/description/  

You are given an integer array nums.  
An integer x is called special if:  
- x appears exactly three times in nums.
- All three occurrences of x are equally spaced in nums. In other words, if all occurrences of x are at indices i1 < i2 < i3, then i2 - i1 = i3 - i2.

Return the number of distinct special integers in nums.  

範例:  
Ex1:  
Input: nums = [1,8,1,5,1,5,8,5]  
Output: 2  
Explanation:  
- 1 is special because it occurs exactly three times at equally spaced indices 0, 2, and 4.
- 5 is special because it occurs exactly three times at equally spaced indices 3, 5, and 7.
- 8 is not special because it occurs only twice.

Therefore, the answer is 2.  

Ex2:  
Input: nums = [8,8,8,8]  
Output: 0  
Explanation:  
8 is not special because it does not occur exactly three times. Therefore, the answer is 0.  

Ex3:  
Input: nums = [8,6,6,8,8]  
Output: 0  
Explanation:  
8 occurs at indices 0, 3, and 4, which are not equally spaced. 6 occurs only twice. Therefore, no integer is special.

* 解題想法:  
首先建立以下的資料結構:
  - count: 用來存放每個數字出現的index
  - d: 用來放每個數字的間距
  - unmatch:用來存放不符合條件的數字有哪些

   接著用迴圈開始檢查該數字是否有出現在unmatch中，如果沒有出現過則接下來檢查是否有出現在count中，如果沒出現過則在count中新增一個空的list，接著檢查在d中是否有之前的間距資料，如果沒有則看墓是否有前一個數字可計算間距並更新間距，如果有的話則檢查是否符合間距，如果間距不符合則將該數放入unmatch中，最後將該數放入count中，迴圈結束後檢查count中的key是否有出現在unmatch中且出現次數是否剛好為三次，統計這些數字就是答案

```python
class Solution:
    def countSpecialIntegers(self, nums: list[int]) -> int:
        res = 0
        count = {}
        unmatch = []
        d = {}
        for i in range(len(nums)):
            n = nums[i]
            if n not in unmatch: 
                if n not in count.keys():
                    count[n] = []
                if count[n]:
                    if n not in d.keys():
                        d[n] = i - count[n][-1]
                    else:
                        if i - count[n][-1] != d[n]:
                            unmatch.append(n)
                count[n].append(i)
        for key, val in count.items():
            if key not in unmatch and len(val) == 3:
                res += 1
        return res
```
