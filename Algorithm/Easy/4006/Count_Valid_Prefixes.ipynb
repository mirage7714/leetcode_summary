# 本題要找出有效的prefix共有多少組 (\#String 
原題目連結: https://leetcode.com/problems/count-valid-prefixes/description/  

You are given a binary string s.  
A prefix of s is considered valid if its characters can be rearranged to form an alternating string.  
Return the number of valid prefixes of s.  
A string is considered alternating if no two adjacent characters are equal.  

範例:  
Ex1:  
Input: s = "00101"  
Output: 3  
Explanation:  
The valid prefixes are:  
- "0": It is already an alternating string.
- "001": It can be rearranged into "010", which is an alternating string.
- "00101": It can be rearranged into "01010", which is an alternating string.

Thus, the answer is 3.  

Ex2:  
Input: s = "101"  
Output: 3  
Explanation:  
All prefixes of s = "101" are already alternating strings. Thus, the answer is 3.

* 解題想法:  
用一個dict來存放目前出現過的每個數字的頻率，接著用迴圈檢查是否符合以下的情形:
- 只出現一個數字，且該數字只出現一次
- 兩個數字都出現過，且兩個數字出現的頻率相等或是相差一

如果符合其中一種條件則將結果加一，最後回傳統計的結果就是答案  

```python
class Solution:
    def countValidPrefixes(self, s: str) -> int:
        res = 0 
        count = {}
        for i in s:
            if i not in count.keys():
                count[i] = 0
            count[i] += 1
            if len(count) == 2 and (max(list(count.values())) == min(list(count.values())) + 1 or max(list(count.values())) == min(list(count.values()))):
                res += 1
            elif len(count) == 1 and max(list(count.values())) == 1:
                res += 1
            print(count)
        return res
```
